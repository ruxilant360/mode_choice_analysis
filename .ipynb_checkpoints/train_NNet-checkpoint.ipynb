```python
def preprocess_data(
    df: pd.DataFrame, target_col: str ):
    """
    Splits df into features/target, auto-detects categorical vs numerical
    columns if not supplied, one-hot encodes categoricals and standard-scales
    numericals.
 
    Pass a fitted `preprocessor` back in (e.g. when transforming a held-out
    test set) to reuse the exact same fitted encoders/scalers instead of
    refitting — this avoids train/test leakage and column mismatches.
 
    Returns
    -------
    X_train : np.ndarray (dense, ready for a Keras model)
    y_train : np.ndarray (raw label array — encode to ints yourself if your
              target isn't already integer-coded, e.g. via LabelEncoder)
    fitted_preprocessor : ColumnTransformer (reuse this on val/test data)
    """
    df = df.copy()
    y = df[target_col].values
    X = df.drop(columns=[target_col])
 
    categorical_features = X.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()
    
    numerical_features = X.select_dtypes(
        include=["int64", "int32", "float64", "float32"]
    ).columns.tolist()

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "cat",
                OneHotEncoder(handle_unknown="ignore", sparse_output=False),
                categorical_features,
            ),
            ("num", StandardScaler(), numerical_features),
        ],
        remainder="drop",
    )
    X_train = preprocessor.fit_transform(X)

    return X_train, y, preprocessor
```

```python
def get_class_weights(y: np.ndarray, method: str = "balanced", beta: float = 0.999) -> dict:
    """
    Computes a class_weight dict for model.fit(class_weight=...).
 
    method="balanced" -> sklearn's standard inverse-frequency weighting
        (n_samples / (n_classes * count_per_class)). This is the common
        industry default and a fine starting point.
 
    method="effective" -> "effective number of samples" re-weighting
        (Cui et al., 2019, "Class-Balanced Loss Based on Effective Number
        of Samples"). This tends to work better than plain inverse-frequency
        when a few classes dominate and the rest are long-tailed, which is
        exactly the 70/30-across-21-classes shape you described — inverse
        frequency alone can overcorrect and blow up gradients for the
        rarest classes.
    """
    classes = np.unique(y)
 
    if method == "balanced":
        weights = compute_class_weight(class_weight="balanced", classes=classes, y=y)
        return {int(c): float(w) for c, w in zip(classes, weights)}
 
    if method == "effective":
        counts = np.array([np.sum(y == c) for c in classes], dtype=np.float64)
        effective_num = 1.0 - np.power(beta, counts)
        weights = (1.0 - beta) / effective_num
        weights = weights / weights.sum() * len(classes)  # normalize, mean ~1
        return {int(c): float(w) for c, w in zip(classes, weights)}
 
    raise ValueError("method must be 'balanced' or 'effective'")
 
```

```python
def train_model(
    X: np.ndarray,
    y: np.ndarray,
    n_splits: int = 5,
    epochs: int = 50,
    batch_size: int = 64,
    learning_rate: float = 1e-3,
    class_weight_method: str = "balanced",
    experiment_name: str = "multiclass_imbalanced_experiment",
    random_state: int = 42,
):
    """
    Runs StratifiedKFold CV, logging one MLflow run per fold (nested under
    a parent run) plus a summary of mean/std metrics across folds.
 
    Class weights are recomputed *inside each fold* from that fold's
    training labels only — never from the full dataset — to avoid leaking
    information about the held-out fold's class distribution.
    """
    y = np.asarray(y)
    num_classes = len(np.unique(y))
    input_dim = X.shape[1]
 
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
 
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog(log_models=False)  # we log models manually per fold
 
    fold_val_acc = []
    fold_val_loss = []
 
    with mlflow.start_run(run_name="cv_parent") as parent_run:
        mlflow.log_params(
            {
                "n_splits": n_splits,
                "epochs": epochs,
                "batch_size": batch_size,
                "learning_rate": learning_rate,
                "class_weight_method": class_weight_method,
                "num_classes": num_classes,
                "input_dim": input_dim,
            }
        )
 
        for fold_idx, (train_idx, val_idx) in enumerate(skf.split(X, y), start=1):
            X_tr, X_val = X[train_idx], X[val_idx]
            y_tr, y_val = y[train_idx], y[val_idx]
 
            class_weights = get_class_weights(y_tr, method=class_weight_method)
 
            with mlflow.start_run(run_name=f"fold_{fold_idx}", nested=True):
                mlflow.log_param("fold", fold_idx)
                mlflow.log_dict(class_weights, f"class_weights_fold_{fold_idx}.json")
 
                model = build_model(input_dim, num_classes, learning_rate)
 
                callbacks = [
                    tf.keras.callbacks.EarlyStopping(
                        monitor="val_loss", patience=8, restore_best_weights=True
                    ),
                    tf.keras.callbacks.ReduceLROnPlateau(
                        monitor="val_loss", factor=0.5, patience=4
                    ),
                ]
 
                history = model.fit(
                    X_tr,
                    y_tr,
                    validation_data=(X_val, y_val),
                    epochs=epochs,
                    batch_size=batch_size,
                    class_weight=class_weights,
                    callbacks=callbacks,
                    verbose=0,
                )
 
                val_loss, val_acc, val_top3 = model.evaluate(X_val, y_val, verbose=0)
                mlflow.log_metrics(
                    {"val_loss": val_loss, "val_accuracy": val_acc, "val_top3_accuracy": val_top3}
                )
                mlflow.tensorflow.log_model(model, artifact_path=f"model_fold_{fold_idx}")
 
                fold_val_acc.append(val_acc)
                fold_val_loss.append(val_loss)
 
        mlflow.log_metrics(
            {
                "mean_val_accuracy": float(np.mean(fold_val_acc)),
                "std_val_accuracy": float(np.std(fold_val_acc)),
                "mean_val_loss": float(np.mean(fold_val_loss)),
                "std_val_loss": float(np.std(fold_val_loss)),
            }
        )
 
    return {
        "fold_val_accuracy": fold_val_acc,
        "fold_val_loss": fold_val_loss,
        "mean_val_accuracy": float(np.mean(fold_val_acc)),
        "mean_val_loss": float(np.mean(fold_val_loss)),
    }
```
