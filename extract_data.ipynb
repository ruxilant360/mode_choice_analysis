```python
import pandas as pd
import numpy as np
```

```python
df_trips = pd.read_csv('trippub.csv', low_memory = False)
df_pers = pd.read_csv('perpub.csv', low_memory = False)
df_hh = pd.read_csv('hhpub.csv', low_memory = False)
```

```python
missing_codes = [-1, -7, -8, -9, -88, -99]
for df in [df_trips, df_pers, df_hh]:
    df.replace(missing_codes, np.nan, inplace=True)
```

```python
df_pers.head(10)
```

```python
hh_cols = ['HOUSEID', 'HHFAMINC', 'HHVEHCNT', 'HHSIZE', 'URBAN', 'HOMEOWN']
```

```python
df_hh_filt = df_hh[hh_cols].copy()
```

```python
per_cols = ['HOUSEID', 'PERSONID', 'R_SEX', 'R_AGE', 'PRMACT', 'WORKER', 'EDUC','PAYPROF', 'DRIVER']
df_per_filt = df_pers[per_cols].copy()
df_per_filt['SEX'] = df_per_filt['R_SEX'].map({1:0, 2:1})
df_per_filt = df_per_filt.drop('R_SEX', axis=1)
```

```python
trip_cols = ['HOUSEID', 'PERSONID', 'TRPTRANS', 'TRVLCMIN', 'TRPMILES', 'DWELTIME', 'WHYTO', 'TRAVDAY' , 'URBRUR']
df_trips_filt = df_trips[trip_cols].copy()
```

```python
df_temp_1 = pd.merge(df_trips_filt, df_per_filt, on=['HOUSEID','PERSONID'], how='left')

trip_data = pd.merge(df_temp_1, df_hh_filt, on = 'HOUSEID', how='left')
```

```python
trip_data.head()
```

```python
print(trip_data.isnull().sum())
```

```python
trip_data.isnull().sum() / len(trip_data) * 100
```

```python
trip_data = trip_data.rename(columns={'DRIVER' : 'CAN-DRIVER'})
```

```python
trip_data.columns
```

```python
trip_data['TRAVDAY'] = trip_data['TRAVDAY'].map({1:'Sunday', 2:'Monday',3:'Tuesday',4:'Wednesday',5:'Thursday',6:'Friday',7:'Saturday'})
```

```python
trip_data.to_csv('trip_data.csv', index=False)
```
